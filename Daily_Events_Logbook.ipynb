```python
import ipywidgets as widgets
from IPython.display import display, clear_output
from datetime import datetime, timedelta

# Widget for event name input
name_input = widgets.Text(
    value='',
    placeholder='Enter event name',
    description='Event:',
    disabled=False
)

# Button to add event
add_button = widgets.Button(
    description='Add Event',
    button_style='success',
    tooltip='Click to add event with current time',
)

output = widgets.Output()

def on_add_event_clicked(b):
    with output:
        clear_output()
        event_name = name_input.value
        if event_name:
            start_time = datetime.now()
            end_time = start_time + timedelta(hours=1)
            add_event(event_name, start_time, end_time)
            print(f"Added event: {event_name} at {start_time.strftime('%Y-%m-%d %H:%M:%S')}")
            display_log()
        else:
            print("Please enter an event name.")

add_button.on_click(on_add_event_clicked)

display(widgets.VBox([name_input, add_button, output]))
```

```python
event_log = []

def add_event(name, start, end):
    """Adds an event to the global event_log list."""
    event_log.append({
        'name': name,
        'start_time': start,
        'end_time': end
    })
```

```python
def display_log():
    """Displays all events currently in the event_log in the output widget."""
    with output:
        if not event_log:
            print("No events added yet.")
        else:
            print("--- Event Log ---")
            for i, event in enumerate(event_log):
                print(f"{i+1}. Name: {event['name']}, Start: {event['start_time'].strftime('%Y-%m-%d %H:%M:%S')}, End: {event['end_time'].strftime('%Y-%m-%d %H:%M:%S')}")
            print("-----------------")
```
